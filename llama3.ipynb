```python
from transformers import AutoTokenizer , AutoModelForCausalLM
import torch

# model_name = "nvidia/Llama3-ChatQA-1.5-8B"
model_name = "meta-llama/Llama-3.1-8B-Instruct"

tokenizer  = AutoTokenizer.from_pretrained(model_name)
model=AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    dtype=torch.float16,
    low_cpu_mem_usage=True
)
```

```python
question_file="/kaggle/input/semeval-12/questions.jsonl"
docs_file="/kaggle/input/semeval-12/docs.json"
```

```python
# questions
questions={}
with open(question_file , "r", encoding="utf-8") as f:
    for line in f:
        q=json.loads(line)
        questions[q['topic_id']]=q
# print(questions)
        # {4: {'topic_id': 4, 
        # 'uuid': 'ab96f31d-3c31-4761-9c4b-66bf5d7bb275',
        # 'target_event': 'Shinzo Abe was pronounced dead at 5:03 p.m.', 
        # 'option_A': 'Former Japanese Prime Minister Shinzo Abe was assassinated at  a campaign rally.', 
        # 'option_B': 'A man fired twice at Shinzo Abe.', 
        # 'option_C': 'Yoshihide Suga succeeded Shinzo Abe in September 2020.',
        # 'option_D': 'Shinzo Abe returned to office in 2012 and later became thelongest-serving prime minister.', 
        # 'golden_answer': 'A,B'},       
```

```python
#docs
docs={}
with open(docs_file, "r", encoding="utf-8") as r:
    docs=json.load(r)

# print(docs)

```

```python
dataset=[]
for entry in docs :
    tid=entry['topic_id']
    if tid in questions:
        q=questions[tid]
        dataset.append({
            "topic_id": tid,
            "event": q["target_event"], 
            "options": {
                "A": q['option_A'],
                "B": q['option_B'],
                "C": q['option_C'],
                "D": q['option_D']
            },
            "answer": q.get("golden_answer"),  
            "context": " ".join([doc["content"] for doc in entry["docs"]])
        })
        
```

```python
import pandas as pd

df=pd.DataFrame(dataset)
# df=df.reset_index()
df.head()
```

```python
def build_chat_messages(row):
    opts = row["options"]
    event = row["event"]
    context = row["context"]
    # print(len(context))
    
    context=context[:1000]
    system_msg = "You are an expert news analyst performing abductive reasoning. Respond ONLY with the letter(s) of the correct answer, separated by commas if more than one. Do NOT include any explanation or extra text."
    user_msg = f"""
    Event: {event}
    Question: {event}
    Context {context}
    Question: {"Which option(s) most plausibly and directly explain the event?"}
Options:
A: {opts['A']}
B: {opts['B']}
C: {opts['C']}
D: {opts['D']}"""
    
    messages = [
        {"role": "system", "content": system_msg},
        {"role": "user", "content": user_msg}
    ]
    return messages

prompts_list = [build_chat_messages(row) for _, row in df.iterrows()]
```

```python
predictions = []
for i in range(0, len(dataset)):
        messages=prompts_list[i]
        event =dataset[i]
    # for messages in prompts_list:
        # print(messages)
    # messages=prompts_list[0]
    
        prompt_text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        
        inputs=tokenizer(prompt_text, return_tensors='pt', padding=True,
            truncation=True,
            max_length=512).to(model.device)
        
        outputs=model.generate(
            **inputs,
               max_new_tokens=2000,
        # temperature=0.9,   # small randomness
        # top_p=0.9,         # nucleus sampling
        do_sample=True,
         pad_token_id=tokenizer.eos_token_id)
        
        decoded=tokenizer.decode(outputs[0],skip_special_tokens=True)
        
        response = decoded.split("assistant")[-1].strip()
        
        # print('---------')
        # print( response) 
        # pred_label = next((c for c in response.upper() if c in ["A", "B", "C", "D"]), "X")
        pred_labels = [c for c in response.upper() if c in ["A", "B", "C", "D"]]
        pred_label = ",".join(pred_labels) if pred_labels else "X"
        # print(pred_label)
        predictions.append({
                "topic_id": event["topic_id"],
                "gold": event["answer"],
                "prediction": pred_label
            })
```

```python
for p in predictions:
    print(p)
```

```python
def evaluate(preds):
    total, score = 0, 0
    for gold, pred in preds:
        if not gold:
            continue
        gold_set, pred_set = set(gold.split(",")), set(pred.split(","))
        if gold_set == pred_set:
            score += 1
        elif gold_set & pred_set:
            score += 0.5
        total += 1
    return score / total if total > 0 else 0
```

```python
eval_input = [(item["gold"], item["prediction"]) for item in predictions]

accuracy = evaluate(eval_input)
print(f"Evaluation score: {accuracy:.4f}")

```
