```python
question_file="/kaggle/input/semeval-12/questions.jsonl"
docs_file="/kaggle/input/semeval-12/docs.json"
```

```python
data=[]
with open(question_file , "r", encoding="utf-8") as f:
    for line in f:
        q=json.loads(line)
        data.append(q)
```

```python
import pandas as pd

df=pd.DataFrame(data)
df=df.reset_index()
df.head()
```

```python
#create prompts 

prompts_list=[]

for index, value in df.iterrows():
    question=value["target_event"]
    answers="option_A: "+value["option_A"]+  " option_B :"+value["option_B"]+" option_C :"+value["option_C"]+" option_D :"+value["option_D"] 
    
    # print(val)
    
    prompts=f"""You are a helpful assistant.Answer with the letters only, separated by commas if needed (e.g., "A" or "A,B") Question : {question}?.  Answers: {answers} .
"""
    prompts_list.append(prompts)


prompts_list[0]
```

```python
inputs =tokanizer(prompts_list[0], return_tensors="pt").to(model.device)
outputs=model.generate(**inputs, max_new_tokens=100)

# print(outputs)

print(tokanizer.decode(outputs[0], skip_special_tokens=True))
```
